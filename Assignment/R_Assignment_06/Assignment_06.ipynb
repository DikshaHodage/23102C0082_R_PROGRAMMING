```r
# Square of numbers 1 to 5
for (i in 1:5) {
  print(i^2)
}
```

```r
# Square of values in a specific vector
for (i in c(2, 4, 6, 7)) {
  print(i^2)
}
```

```r
# Counting elements where xval / 2 > 3
x <- c(2, 4, 6, 8, 10, 12)

excount <- function(x) {
  count <- 0
  for (xval in x) {
    if (xval / 2 > 3) {
      count <- count + 1
    }
  }
  print(count)
}

# Run the function
excount(x)
```

```r
#  Nested looping
child <- c("child1", "child2", "child3")
sweet <- c("sweet1", "sweet2", "sweet3")

for (x in child) {
  for (y in sweet) {
    print(paste(x, y))
  }
}
```

```r
# Break loop when "tea" is encountered
drink <- c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    break
  }
  print(x)
}
```

```r
#  Skip lemonade using next
drink <- c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "lemonade") {
    next
  }
  print(x)
}
```

```r
#  Skip tea using next
for (x in drink) {
  if (x == "tea") {
    next
  }
  print(x)
}
```

```r
#  while loop for i < 10
i <- 1
while (i < 10) {
  print(i^2)
  i <- i + 2
}
```

```r
#  Sum function with while loop
sumfunction <- function(start_num = 22) {
  sum_val <- 0
  number <- start_num
  while (number <= 25) {
    sum_val <- sum_val + number
    number <- number + 1
  }
  print(paste("The sum of numbers received from the While Loop: ", sum_val))
}

# Execute function starting at 22 (22 + 23 + 24 + 25 = 94)
sumfunction(22)

# Verification check
print(22 + 23 + 24 + 25)
```

```r
#  repeat loop printing i^2 until i > 10
i <- 1
repeat {
  print(i^2)
  i <- i + 2
  if (i > 10) {
    break
  }
}
```

```r
# repeat loop with next and break
i <- 1
repeat {
  i <- i + 1
  if (i < 10) next
  print(i^2)
  if (i >= 13) break
}
```

```r
#  Single variable function
abc <- function(x) {
  x^2
}

# Calling function with different arguments
print(abc(3))
print(abc(6))
print(abc(9))
```

```r
#  Two variable function
abc_two <- function(x, y) {
  x^2 + y^2
}

# Calling function with different pairs
print(abc_two(3, 4))
print(abc_two(10, 10))
print(abc_two(-2, -3))
```

```r
# Trigonometric calculation function
abc_trig <- function(x) {
  sin(x)^2 + cos(x)^2 + x
}

# Calling function
print(abc_trig(9))
print(abc_trig(99))
print(abc_trig(-15))
```

```r
#  Function without arguments
abc_no_arg <- function() {
  for (i in 1:3) {
    print(i^3)
  }
}

# Executing function
abc_no_arg()
```

```r
# Default step size (+1 or -1)
print(seq(from = 2, to = 4))
print(seq(from = 4, to = 2))
print(seq(from = -4, to = 4))
```

```r
# Sequence from 10 to 20 with step 2
seq(from = 10, to = 20, by = 2)
```

```r
# Sequence from 20 to 10 with step -2
seq(from = 20, to = 10, by = -2)
```

```r
# Sequence from 3 to -2 with step -0.5
seq(from = 3, to = -2, by = -0.5)
```

```r
# Sequence ending at 10 with length 10
seq(to = 10, length.out = 10)
```

```r
# Sequence starting from 10 with length 10
seq(from = 10, length.out = 10)
```

```r
# Sequence from 10, length 10, step 0.1
seq(from = 10, length.out = 10, by = 0.1)
```

```r
# Sequence from 10, length 10, step -2
seq(from = 10, length.out = 10, by = -2)
```

```r
# Sequence from 10, length 5, step -0.2
seq(from = 10, length.out = 5, by = -0.2)
```
