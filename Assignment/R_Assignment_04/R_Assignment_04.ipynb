## 1. Matrix creation, `byrow` and indexing

The important idea is that R fills a matrix column-wise by default. Using `byrow = TRUE` changes this behavior.

```r
# Create the same data in two different layouts
x_col <- matrix(1:12, nrow = 4, ncol = 3)
x_row <- matrix(1:12, nrow = 4, ncol = 3, byrow = TRUE)

x_col
x_row

# Access one row, one column and a submatrix
x_row[3, ]
x_row[, 2]
x_row[2:4, 1:2]
```

## 2. Row and column names

Use `rownames()` and `colnames()` to label matrix dimensions.

```r
x <- matrix(1:12, nrow = 4, ncol = 3)

rownames(x) <- paste0("r", 1:4)
colnames(x) <- paste0("c", 1:3)

x
```

## 3. Matrix indexing with selected rows and columns

This combines the `[]` indexing operator with `c()` to select non-consecutive rows/columns.

```r
x <- matrix(1:15, nrow = 5, ncol = 3, byrow = TRUE)

# Rows 1 and 4, columns 1 and 3
x[c(1, 4), c(1, 3)]

# Rows 4 and 5, columns 2 and 3
x[4:5, 2:3]
```

## 4. Diagonal matrix and identity matrix

`diag()` can create an identity matrix or a diagonal matrix with a specified diagonal value.

```r
I3 <- diag(1, nrow = 3, ncol = 3)
D3 <- diag(5, nrow = 3, ncol = 3)

I3
D3
```

## 5. Transpose and row/column summaries

`t()` changes rows into columns. `rowSums()`, `colSums()`, `rowMeans()` and `colMeans()` summarize matrix dimensions.

```r
x <- matrix(1:8, nrow = 4, ncol = 2, byrow = TRUE)

x
t(x)

rowSums(x)
colSums(x)

rowMeans(x)
colMeans(x)
```

## 6. Scalar arithmetic: `+`, `-`, `*`, `/`

A scalar operation is applied to every element of the matrix.

```r
x <- matrix(1:8, nrow = 4, ncol = 2, byrow = TRUE)

x + 5       # addition
x - 5       # subtraction
5 * x       # multiplication
x / 2       # division
```

## 7. Matrix addition and subtraction

For ordinary matrix `+` and `-`, the matrices must have compatible dimensions. The operation is element-wise.

```r
x <- matrix(1:8, nrow = 4, ncol = 2, byrow = TRUE)
y <- matrix(11:18, nrow = 4, ncol = 2, byrow = TRUE)

x + y
x - y

# A useful combined expression from the lecture idea
x + 4*x
4*x - x
```

## 8. `%*%` – true matrix multiplication

`%*%` is different from `*`. It performs matrix multiplication using rows of the first matrix and columns of the second.

```r
x <- matrix(1:8, nrow = 4, ncol = 2, byrow = TRUE)
y <- matrix(11:18, nrow = 2, ncol = 4, byrow = TRUE)

x %*% y
y %*% x

# Compare element-wise multiplication with matrix multiplication
a <- matrix(c(1, 2, 3, 4), nrow = 2, byrow = TRUE)
b <- matrix(c(5, 6, 7, 8), nrow = 2, byrow = TRUE)

a * b
a %*% b
```

## 9. Transpose multiplied by the original matrix

A common matrix calculation is X'X. In R this is written as `t(x) %*% x`.

```r
x <- matrix(1:8, nrow = 4, ncol = 2, byrow = TRUE)

xtx <- t(x) %*% x
xxt <- x %*% t(x)

xtx
xxt
```

## 10. `crossprod()` for X'X

The lecture uses `crossprod(x)` as a faster way to calculate `t(x) %*% x`.

```r
x <- matrix(1:8, nrow = 4, ncol = 2, byrow = TRUE)

t(x) %*% x
crossprod(x)

# They should give the same matrix
all(t(x) %*% x == crossprod(x))
```

## 11. Inverse using `solve()`

For an invertible square matrix, `solve()` can calculate the inverse.

```r
y <- matrix(c(84, 100,
              100, 120),
            nrow = 2, byrow = TRUE)

y_inv <- solve(y)
y_inv

# Check the inverse concept
y %*% y_inv
```

## 12. Eigen values and eigen vectors

`eigen()` returns both eigen values and eigen vectors of a square matrix.

```r
y <- matrix(c(84, 100,
              100, 120),
            nrow = 2, byrow = TRUE)

eig <- eigen(y)

eig$values
eig$vectors
```

# Relational and Logical Operators

The Week 4 notes cover `>`, `>=`, `<`, `<=`, `==`, `!=`, `!`, `&`, `&&`, `|`, `||`, `xor()`, `isTRUE()` and `isFALSE()`.

The following examples are deliberately grouped so that each example demonstrates a concept rather than merely testing one number.

## 13. Relational operators together

Relational operators produce logical values (`TRUE` or `FALSE`).

```r
marks <- c(45, 72, 81, 60, 39)

marks > 60
marks >= 60
marks < 60
marks <= 60
marks == 60
marks != 60
```

## 14. Comparing two vectors element-by-element

Relational operators work element-wise on vectors.

```r
x <- c(1, 2, 3)
y <- c(4, 2, 1)

x > y
x < y
x == y
x != y
```

## 15. Negation `!`

`!` reverses a logical result: `TRUE` becomes `FALSE` and vice versa.

```r
passed <- c(TRUE, FALSE, TRUE, FALSE)

passed
!passed

# Negating a condition
marks <- c(45, 72, 81, 60, 39)
marks > 60
!(marks > 60)
```

## 16. Element-wise AND `&`

`&` evaluates the condition for every element of a vector.

```r
marks <- c(45, 72, 81, 60, 39)

eligible <- (marks >= 50) & (marks <= 80)

marks >= 50
marks <= 80
eligible

# Use the logical result to select values
marks[eligible]
```

## 17. Element-wise OR `|`

`|` evaluates both conditions element-by-element.

```r
marks <- c(45, 72, 81, 60, 39)

special <- (marks < 50) | (marks > 80)

special
marks[special]
```

## 18. Why `&` and `&&` are different

`&` works element-wise. `&&` examines only the first element of the operands.

This is one of the important conceptual distinctions in the Week 4 notes.

```r
x <- c(8, 18)

(x < 10) & (x > 2)
# The error happens because && expects single logical values.
# When (x < 10) evaluates to c(TRUE, FALSE), it's a vector of length 2.
# && tries to use this vector where a single logical is expected, leading to the 'length = 2' error.
# To use && correctly with vectors, you should explicitly select the first element.
(x[1] < 10) && (x[1] > 2)

# The first result contains one logical value for each element.
# The second result is based only on the first element.
```

## 19. Why `|` and `||` are different

Similarly, `|` is element-wise while `||` evaluates only the first element.

```r
x <- c(8, 18)

(x < 10) | (x < 2)
# The error happens because || expects single logical values, similar to &&.
# When (x < 10) evaluates to c(TRUE, FALSE), it's a vector of length 2.
# || tries to use this vector where a single logical is expected, leading to the 'length = 2' error.
# To use || correctly with vectors, you should explicitly select the first element.
(x[1] < 10) || (x[1] < 2)
```

## 20. Short-circuit behavior with `&&`

`&&` evaluates from left to right and stops once the final result is determined.

```r
x <- 1:6

# Only the first element of x is considered by &&
# The error happened because && expects single logical values, not vectors.
# (x > 2) evaluates to a vector of length 6, causing the 'length = 6' error.
# To use && correctly, you must explicitly refer to the first element.
(x[1] > 2) && (x[1] < 5)

# Equivalent first-element view
(x[1] > 2) & (x[1] < 5)
```

## 21. Combining relational and logical operators for filtering

This is a practical example of using several operators together.

```r
x <- 1:10

# Values greater than 2 AND less than 8
x[(x > 2) & (x < 8)]

# Values less than 3 OR greater than 8
x[(x < 3) | (x > 8)]
```

## 22. `xor()` – exclusive OR

`xor()` is TRUE when exactly one of the two logical inputs is TRUE. It is FALSE when both are TRUE or both are FALSE.

```r
a <- c(TRUE, TRUE, FALSE, FALSE)
b <- c(TRUE, FALSE, TRUE, FALSE)

xor(a, b)

# Compare ordinary OR and exclusive OR
a | b
xor(a, b)
```

## 23. `isTRUE()` and `isFALSE()`

These functions test whether an expression evaluates to the corresponding logical value.

```r
isTRUE(8 > 6)
isTRUE(8 < 6)

isFALSE(5 > 8)
isFALSE(5 < 8)

# Combined with an expression
x <- 5
isTRUE(2*x > 11)
isTRUE(3*x < 20)
```

## 24. Logical values produced from arithmetic expressions

Arithmetic and relational operators can be combined. The result can then be checked with `is.logical()`.

```r
x <- 5

Logical4 <- (2*x > 11)
Logical5 <- (3*x < 20)

Logical4
Logical5

is.logical(Logical4)
is.logical(Logical5)
```

## 25. Conceptual mini-problem: classify values using several operators

This combines relational operators, `&`, `|`, `!` and logical indexing in one example.

```r
scores <- c(35, 48, 55, 67, 74, 89, 92)

# Passing range
pass <- scores >= 50

# Distinction range
distinction <- scores >= 75

# Students who passed but are not in distinction
regular_pass <- pass & !distinction

# Very low OR very high scores
extreme <- (scores < 40) | (scores > 90)

list(
  pass = scores[pass],
  regular_pass = scores[regular_pass],
  extreme = scores[extreme]
)
```

## 26. Conceptual mini-problem: matrix + logical operators

This combines matrix operations with relational/logical filtering.

```r
m <- matrix(1:12, nrow = 4, ncol = 3, byrow = TRUE)

m

# Select all matrix elements greater than 4 and less than 10
m[(m > 4) & (m < 10)]

# Select elements that are <= 2 OR >= 11
m[(m <= 2) | (m >= 11)]

# Negate a condition
m[!(m > 4)]
```
